```python
import numpy as np
```

```python
bit_map:np.ndarray = np.load('2D_bit_map.npy')

ox = []
oy = []
for row, row_bits in enumerate(bit_map):
    for col, bit in enumerate(row_bits):
        if bit:
            ox.append(col)
            oy.append(row)
```
