```python
import numpy as np
from path_planning import AStarPlanner
import matplotlib.pyplot as plt
```

```python
BITMAP_PATH = '/home/dtu_dev/ws/pico_ws/src/waypoint_navigation/map/2D_bit_map.npy'
bit_map:np.ndarray = np.load(BITMAP_PATH)

ox = []
oy = []
for row, row_bits in enumerate(bit_map):
    for col, bit in enumerate(row_bits):
        if bit:
            ox.append(col)
            oy.append(row)
```

```python
show_animation = False
# start and goal position
sx = 10.0
sy = 10.0
gx = 50.0
gy = 50.0
grid_size = 2.0
robot_radius = 1.0

plt.plot(ox, oy, ".k")
plt.plot(sx, sy, "og")
plt.plot(gx, gy, "xb")
plt.grid(True)
plt.axis("equal")

a_star = AStarPlanner(ox, oy, grid_size, robot_radius)
rx, ry = a_star.planning(sx, sy, gx, gy)

plt.plot(rx, ry, "-r")
plt.pause(0.001)
plt.show()
```
